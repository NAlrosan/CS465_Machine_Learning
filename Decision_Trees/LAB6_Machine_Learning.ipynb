```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
from scipy.stats import zscore

# Step 1: Load Dataset
df = pd.read_csv("diabetesDataset.csv")

# Step 2: Clean Column Names
df.columns = [
    "Pregnancies", "Glucose", "BloodPressure", "SkinThickness", "Insulin",
    "BMI", "DiabetesPedigree", "Age", "Diabetes"
]

# Step 3: Handle Missing Values & Outliers
df.replace(0, np.nan, inplace=True)
df.fillna(df.median(), inplace=True)  # Fill zeros (or missing) with median

# Outlier Removal using Z-score
df = df[(np.abs(zscore(df.select_dtypes(include=np.number))) < 3).all(axis=1)]

# Step 4: Encode Target Column if needed
if df['Diabetes'].dtype == 'object':
    df['Diabetes'] = LabelEncoder().fit_transform(df['Diabetes'])

# Step 5: Feature Scaling
X = df.drop("Diabetes", axis=1)
y = df["Diabetes"]
X_scaled = StandardScaler().fit_transform(X)

# Step 6: Split the Dataset
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Step 7: Decision Tree
dt_model = DecisionTreeClassifier()
dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)

# Step 8: Random Forest
rf_model = RandomForestClassifier()
rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)

# Step 9: XGBoost
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
xgb_model.fit(X_train, y_train)
xgb_pred = xgb_model.predict(X_test)

# Step 10: Compare Performance
print("Decision Tree Accuracy:", accuracy_score(y_test, dt_pred))
print("Random Forest Accuracy:", accuracy_score(y_test, rf_pred))
print("XGBoost Accuracy:", accuracy_score(y_test, xgb_pred))
```
